## Google Gemini Model Intregration

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    temperature=0.2,
)

response = model.invoke("Explain LangChain in simple words.")
print(response.content)
```

## GROQ Model Intregration

```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

# Load API key
load_dotenv()

# Initialize Groq model
model = init_chat_model(
    model="openai/gpt-oss-20b",
    model_provider="groq",
    temperature=0.7,
)

# Generate response
response = model.invoke("Why do parrots talk?")

print(response.content)
```

## Streaming in LangChain

### Streaming is a technique in which an application receives and displays a model’s response in smaller chunks instead of waiting for the entire response to finish.

```python
model.invoke("Write me a 200 words paragraph on Artifical Intelligence")
```

### How does streaming work?

### The process generally follows these steps:

Step 1 — User sends a question

"What is machine learning?"

Step 2 — LangChain calls the model

Sends the prompt to the LLM provider

Step 3 — Model returns chunks

Small pieces of the generated response

Step 4 — Application displays each chunk

The answer grows as new chunks arrive

```python
for chunk in model.stream(
    "Explain machine learning in simple English."
):
    print(chunk.content, end="", flush=True)

print()
```

## Batch Processing

### Batch processing means sending multiple inputs to a model or chain through a single method call and receiving a result for each input.

```python
questions = [
    "What is Python?",
    "What is Machine Learning?",
    "What is Deep Learning?",
    "What is Generative AI?",
    "What is LangChain?",
]

responses = model.batch(questions)

for i, (question, response) in enumerate(
    zip(questions, responses),
    start=1,
):
    print(f"\nQuestion {i}: {question}")
    print(f"Answer: {response.content}")
```
