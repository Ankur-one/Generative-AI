```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    temperature=0.2,
)
model
```

```python
model.invoke("Please tell What is artifical intelligence ?")
```

## Text Prompts

```python
model.invoke("What is langchain")
```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage("You are a poetry expert"),
    HumanMessage("Write a poem on artifical intelligence")
]

response = model.invoke(messages)
print(response.content)
```

```python
system_msg = SystemMessage("You are a helpful coading assistant.")

messages = [
    system_msg,
    HumanMessage("How do I create a REST API?")
]
response = model.invoke(messages)
print(response)
```

### Detailed info to the LLM through System message

```python
from langchain.messages import SystemMessage, HumanMessage

system_msg = SystemMessage("""
You are a senior developer with expertise in web frameworks.
Always provide code examples and explain your reasoning.
Be concise but through in your explanation.
""")

messages = [
    system_msg,
    HumanMessage("How do I create a REST API?")
]

response = model.invoke(messages)
print(response.content)
```

```python
## Message Metadata
human_msg = HumanMessage(
    content="Hello",
    name='alice',
    id="msg_123",
)

response = model.invoke([
    human_msg
])
print(response)
```

```python
from langchain.messages import AIMessage,HumanMessage,SystemMessage

ai_msg = AIMessage("I'd be happy to help you with that question!")

messages = [
    SystemMessage("You are a helpful assistant"),
    HumanMessage("Can you help me?"),
    ai_msg,
    HumanMessage("Great What's 2+2?")
]

response = model.invoke(messages)
print(response.content)
```

```python
response.usage_metadata
```

```python
from langchain.messages import AIMessage,ToolMessage

ai_message = AIMessage(
    content=[],
    tool_calls=[{
        "name": "get_weather",
        "args": {"loaction": "Malout"},
        "id": "call_123"
    }]
)

weather_result = "Sunny, 72^o F",
tool_message = ToolMessage(
    content=weather_result,
    tool_call_id="call_123"
)

messages = [
    HumanMessage("What is the weather in Malout"),
    ai_message,
    tool_message
]

response = model.invoke(messages)
print(response)
```
